### Lọc bỏ các mẫu điện thoại cơ bản (máy cục gạch) và thiết bị không phải smartphone
Chúng ta sẽ nhận diện và loại bỏ các mẫu máy cục gạch dựa trên đặc điểm dung lượng bộ nhớ RAM hoặc ROM hiển thị dưới đơn vị `MB` thay vì `GB` (hoặc không có cả Hệ điều hành, CPU và RAM), đồng thời lọc bỏ các mẫu máy tính bảng (tablet), đồng hồ thông minh (smartwatch) và các dòng bị lỗi đơn vị tiền tệ để tránh ảnh hưởng đến các phân tích sau này.

```python
import re
import pandas as pd

def is_non_target_smartphone(row):
    ram_str = str(row['RAM']).upper() if pd.notna(row['RAM']) else ''
    rom_str = str(row['internal_memory']).upper() if pd.notna(row['internal_memory']) else ''

    # 1. Lọc Feature Phone (Máy cục gạch)
    if 'MB' in ram_str or 'MB' in rom_str or '512 ROM' in rom_str:
        return True
    if pd.isna(row['OS']) and pd.isna(row['CPU']) and pd.isna(row['RAM']):
        return True

    # 2. Lọc Smartwatch, Tablet, Không hỗ trợ SIM
    os_str = str(row['OS']).upper() if pd.notna(row['OS']) else ''
    model_str = str(row['model']).upper()

    if 'WATCH' in os_str or 'WATCH' in model_str:
        return True
    if str(row['network_technology']).strip().upper() == 'NO CELLULAR CONNECTIVITY':
        return True
    if any(k in model_str for k in ['IPAD', 'TAB ', 'TAB-', 'TABLET', 'TOUGHPAD', 'FUNBOOK', 'STREAK', 'SPRINGBOARD']):
        return True

    # 3. Kích thước màn hình: Quét cả 2 cột vì dữ liệu bị nhầm lẫn
    res_str = str(row['display_resolution']) if pd.notna(row['display_resolution']) else ''
    size_str = str(row['display_size']) if pd.notna(row['display_size']) else ''
    combined_display_info = f"{res_str} {size_str}"

    m = re.search(r'(\d+(?:\.\d+)?)\s*inches', combined_display_info, re.IGNORECASE)
    if m:
        size_inch = float(m.group(1))
        # Vẫn nới ngưỡng lên 8.0 để giữ lại smartphone gập (VD: Z Fold 7.6 inch)
        if size_inch > 8.0 or size_inch < 3.0:
            return True

    # 4. Lọc Outlier giá và điện thoại thủ công
    if row['approx_price_EUR'] > 2500 or str(row['brand']).strip().upper() == 'VERTU':
        return True

    return False

# Áp dụng bộ lọc
initial_count = len(df)
df = df[~df.apply(is_non_target_smartphone, axis=1)].copy()

print(f"Số lượng thiết bị ban đầu: {initial_count}")
print(f"Số lượng máy sau khi lọc: {len(df)}")
```

### 1. Xử lý cột `RAM` và `internal_memory` để tách lấy RAM và ROM

```python
import re
import pandas as pd
import numpy as np

def extract_ram_rom(row):
    ram_val = row['RAM']
    rom_val = row['internal_memory']

    # Tách lấy dung lượng RAM (ví dụ: "2 GB RAM", "3/4 GB RAM")
    ram = None
    if pd.notna(ram_val):
        s_ram = str(ram_val).upper()
        m_ram = re.findall(r'(?<![A-Z0-9])(\d+(?:\.\d+)?)\s*GB\s*RAM', s_ram)
        if not m_ram:
            m_ram = re.findall(r'(?<![A-Z0-9])(\d+(?:\.\d+)?)\s*GB', s_ram)
        vals = [float(x) for x in m_ram if float(x) <= 16]
        if vals:
            ram = f"{vals[0]} GB"

    # Tách lấy dung lượng bộ nhớ trong ROM (ví dụ: "16 GB", "16/32 GB")
    rom = None
    if pd.notna(rom_val):
        s_rom = str(rom_val).upper().strip()
        m_rom = re.findall(r'(?<![A-Z0-9])(\d+(?:\.\d+)?)\s*GB', s_rom)
        if m_rom:
            rom = f"{float(m_rom[0])} GB"
        elif re.match(r'^\d+(?:\.\d+)?$', s_rom):
            rom = f"{float(s_rom)} GB"

    return ram, rom

# Áp dụng hàm để chuẩn hóa cột RAM và tạo cột ROM
ram_rom_extracted = df.apply(extract_ram_rom, axis=1)
df['RAM_Clean'] = [x[0] for x in ram_rom_extracted]
df['ROM'] = [x[1] for x in ram_rom_extracted]

display(df[['RAM', 'internal_memory', 'RAM_Clean', 'ROM']].head())
```

### 2. Xử lý cụm cột Màn hình (`display_resolution`, `display_type`, `display_size`) để tách Kích thước, Tỷ lệ màn hình, Tấm nền và Độ phân giải
*(Lưu ý: Trong bộ dữ liệu gốc, hai cột `display_resolution` và `display_size` bị đặt ngược tên nhau — cột `display_resolution` chứa kích thước inch kèm tỷ lệ màn hình, còn `display_size` lại chứa độ phân giải pixels).*

```python
def extract_screen_info(row):
    res_col = str(row['display_resolution']) if pd.notna(row['display_resolution']) else ""
    size_col = str(row['display_size']) if pd.notna(row['display_size']) else ""
    type_col = str(row['display_type']) if pd.notna(row['display_type']) else ""

    # 1. Kích thước màn hình: đi kèm chữ "inches"
    size_match = re.search(r'(\d+(?:\.\d+)?)\s*inches', res_col, re.IGNORECASE)
    size = f"{size_match.group(1)} inch" if size_match else None

    # 2. Tỷ lệ màn hình so với thân máy (screen-to-body ratio): đi kèm ký hiệu "%"
    ratio_match = re.search(r'(\d+(?:\.\d+)?)\s*%', res_col)
    screen_ratio = f"{ratio_match.group(1)} %" if ratio_match else None

    # 3. Tấm nền (Công nghệ màn hình): Super AMOLED, AMOLED, IPS LCD, Super LCD, TFT LCD...
    t_up = type_col.upper()
    if 'SUPER AMOLED' in t_up:
        panel = 'Super AMOLED'
    elif 'AMOLED' in t_up:
        panel = 'AMOLED'
    elif 'OLED' in t_up:
        panel = 'OLED'
    elif 'IPS' in t_up:
        panel = 'IPS LCD'
    elif 'SUPER LCD' in t_up or 'S-LCD' in t_up:
        panel = 'Super LCD'
    elif 'TFT' in t_up:
        panel = 'TFT LCD'
    elif 'PLS' in t_up:
        panel = 'PLS LCD'
    else:
        panel = None

    # 4. Độ phân giải: ví dụ 720 x 1280 pixels
    res_match = re.search(r'(\d+\s*[xX]\s*\d+\s*(?:Pixels|pixels)?)', size_col)
    resolution = res_match.group(1).strip() if res_match else None

    return size, screen_ratio, panel, resolution

# Áp dụng hàm tách thông tin màn hình
screen_extracted = df.apply(extract_screen_info, axis=1)
df['Kích Thước Màn Hình'] = [x[0] for x in screen_extracted]
df['Tỷ Lệ Màn Hình'] = [x[1] for x in screen_extracted]
df['Tấm Nền Màn Hình'] = [x[2] for x in screen_extracted]
df['Độ Phân Giải Màn Hình'] = [x[3] for x in screen_extracted]

display(df[['display_resolution', 'display_type', 'Kích Thước Màn Hình', 'Tỷ Lệ Màn Hình', 'Tấm Nền Màn Hình', 'Độ Phân Giải Màn Hình']].head(10))
```

### 3. Xử lý cột `primary_camera` và `secondary_camera` để lấy độ phân giải Camera sau và Camera trước (MP)

```python
def extract_camera_info(row):
    p_str = str(row['primary_camera']).upper() if pd.notna(row['primary_camera']) else ""
    s_str = str(row['secondary_camera']).upper() if pd.notna(row['secondary_camera']) else ""

    # Tách độ phân giải Camera sau lớn nhất
    mp_rear = re.findall(r'(?<![A-Z\d])(\d+(?:\.\d+)?)\s*MP', p_str)
    vals_p = [float(val) for val in mp_rear if float(val) <= 100]
    if vals_p:
        rear_cam = f"{max(vals_p)} MP"
    elif 'VGA' in p_str:
        rear_cam = "0.3 MP"
    else:
        rear_cam = None

    # Tách độ phân giải Camera trước (nếu ghi 'No' thì gán 0 MP, 'VGA' gán 0.3 MP)
    if s_str.strip() == 'NO':
        front_cam = "0 MP"
    else:
        mp_front = re.findall(r'(?<![A-Z\d])(\d+(?:\.\d+)?)\s*MP', s_str)
        vals_s = [float(val) for val in mp_front if float(val) <= 100]
        if vals_s:
            front_cam = f"{max(vals_s)} MP"
        elif 'VGA' in s_str:
            front_cam = "0.3 MP"
        else:
            front_cam = None

    return rear_cam, front_cam

# Áp dụng hàm để lấy độ phân giải camera sau và camera trước
camera_extracted = df.apply(extract_camera_info, axis=1)
df['Camera Sau'] = [x[0] for x in camera_extracted]
df['Camera Trước'] = [x[1] for x in camera_extracted]

display(df[['primary_camera', 'secondary_camera', 'Camera Sau', 'Camera Trước']].head(10))
```

### 4. Xử lý cột `battery` để lấy dung lượng Pin (mAh) và Thiết kế Pin (Nguyên khối / Tháo rời)

```python
def extract_battery_info(text):
    if pd.isna(text):
        return None, None
    text_str = str(text)

    # Tìm dung lượng pin (ví dụ: 3400 mAh)
    battery_match = re.search(r'(\d+)\s*mAh', text_str, re.IGNORECASE)
    battery_capacity = f"{battery_match.group(1)} mAh" if battery_match else None

    # Xác định thiết kế pin liền máy (Non-removable) hay tháo rời (Removable)
    if 'NON-REMOVABLE' in text_str.upper():
        battery_design = 'Nguyên khối'
    else:
        battery_design = 'Tháo rời'

    return battery_capacity, battery_design

# Áp dụng hàm để lấy dung lượng pin và kiểu thiết kế pin
battery_extracted = df['battery'].apply(extract_battery_info)
df['Dung Lượng Pin'] = [x[0] for x in battery_extracted]
df['Thiết Kế Pin'] = [x[1] for x in battery_extracted]

display(df[['battery', 'Dung Lượng Pin', 'Thiết Kế Pin']].head(10))
```

### 5. Xử lý thông số kích thước (`dimentions`), trọng lượng (`weight_g`) và kiểm tra tính năng Cảm biến vân tay (`sensors`)

```python
def extract_design_and_sensors(row):
    dim_text = str(row['dimentions']) if pd.notna(row['dimentions']) else ""
    sensors_text = str(row['sensors']) if pd.notna(row['sensors']) else ""

    # 1. Tách độ dày thân máy (mm) từ chuỗi kích thước Dài x Rộng x Dày mm
    thick_match = re.search(r'x\s*\d+(?:\.\d+)?\s*x\s*(\d+(?:\.\d+)?)\s*mm', dim_text, re.IGNORECASE)
    if not thick_match:
        thick_match = re.search(r'(\d+(?:\.\d+)?)\s*mm\s*thickness', dim_text, re.IGNORECASE)
    thickness = f"{thick_match.group(1)} mm" if thick_match else None

    # 2. Chuẩn hóa trọng lượng máy (g)
    w_val = pd.to_numeric(row['weight_g'], errors='coerce')
    weight = f"{w_val} g" if pd.notna(w_val) else None

    # 3. Kiểm tra trang bị cảm biến vân tay (Fingerprint)
    if re.search(r'fingerprint', sensors_text, re.IGNORECASE):
        fingerprint = "Có"
    else:
        fingerprint = "Không"

    return thickness, weight, fingerprint

# Áp dụng hàm để lấy độ dày, trọng lượng và cảm biến vân tay
design_extracted = df.apply(extract_design_and_sensors, axis=1)
df['Độ Dày'] = [x[0] for x in design_extracted]
df['Trọng Lượng'] = [x[1] for x in design_extracted]
df['Cảm Biến Vân Tay'] = [x[2] for x in design_extracted]

display(df[['dimentions', 'weight_g', 'sensors', 'Độ Dày', 'Trọng Lượng', 'Cảm Biến Vân Tay']].head(10))
```

### 6. Kiểm tra hỗ trợ kết nối mạng di động (2G, 3G hoặc 4G LTE)

```python
def check_network_support(row):
    net_tech = str(row['network_technology']) if pd.notna(row['network_technology']) else ""
    bands_4g = str(row['4G_bands']) if pd.notna(row['4G_bands']) else ""
    combined_text = (net_tech + " " + bands_4g).upper()

    # Kiểm tra sự xuất hiện của chuẩn mạng cao nhất mà máy hỗ trợ
    if "LTE" in combined_text or "4G" in combined_text:
        return "4G / LTE"
    elif any(k in combined_text for k in ["HSPA", "3G", "UMTS", "EVDO"]):
        return "3G"
    else:
        return "2G"

# Áp dụng hàm để xác định loại mạng hỗ trợ
df['Hỗ Trợ Mạng'] = df.apply(check_network_support, axis=1)

display(df[['network_technology', '4G_bands', 'Hỗ Trợ Mạng']].head(10))
```

### 7. Trích xuất Hãng Thiết kế Chip, Số Nhân CPU và Xung Nhịp CPU

```python
def extract_chip_info(row):
    cpu_str = str(row['CPU']).strip() if pd.notna(row['CPU']) else ""
    chip_str = str(row['Chipset']).strip() if pd.notna(row['Chipset']) else ""
    combined_upper = (chip_str + " " + cpu_str).upper()

    # 1. Trích xuất hãng chip (Brand)
    if "SNAPDRAGON" in combined_upper or "QUALCOMM" in combined_upper:
        brand = "Snapdragon"
    elif any(k in combined_upper for k in ["MEDIATEK", "HELIO", "MT6", "MT8"]):
        brand = "MediaTek"
    elif "EXYNOS" in combined_upper:
        brand = "Exynos"
    elif "APPLE" in combined_upper or str(row['brand']).strip().upper() == "APPLE":
        brand = "Apple"
    elif "KIRIN" in combined_upper or "HISILICON" in combined_upper:
        brand = "Kirin"
    elif "SPREADTRUM" in combined_upper:
        brand = "Spreadtrum"
    elif "INTEL" in combined_upper or "ATOM" in combined_upper:
        brand = "Intel"
    elif pd.isna(row['Chipset']):
        brand = None
    else:
        brand = "Khác"

    # 2. Trích xuất số nhân CPU (Cores)
    cpu_upper = cpu_str.upper()
    if not cpu_str:
        cores = None
    elif "DECA-CORE" in cpu_upper or "10-CORE" in cpu_upper:
        cores = "10 nhân"
    elif "OCTA-CORE" in cpu_upper or "8-CORE" in cpu_upper:
        cores = "8 nhân"
    elif "HEXA-CORE" in cpu_upper or "6-CORE" in cpu_upper:
        cores = "6 nhân"
    elif "QUAD-CORE" in cpu_upper or "4-CORE" in cpu_upper:
        cores = "4 nhân"
    elif "DUAL-CORE" in cpu_upper or "2-CORE" in cpu_upper:
        cores = "2 nhân"
    else:
        cores = "1 nhân"

    # 3. Trích xuất xung nhịp tối đa (GHz)
    ghz_matches = re.findall(r'(\d+(?:\.\d+)?)\s*GHz', cpu_str, re.IGNORECASE)
    if ghz_matches:
        clock = f"{max([float(x) for x in ghz_matches])} GHz"
    else:
        mhz_matches = re.findall(r'(\d+(?:\.\d+)?)\s*MHz', cpu_str, re.IGNORECASE)
        clock = f"{round(max([float(x) for x in mhz_matches]) / 1000.0, 2)} GHz" if mhz_matches else None

    return brand, cores, clock

# Áp dụng hàm để tạo 3 cột thông số vi xử lý
chip_extracted = df.apply(extract_chip_info, axis=1)
df['Hãng Chip'] = [x[0] for x in chip_extracted]
df['Số Nhân CPU'] = [x[1] for x in chip_extracted]
df['Xung Nhịp CPU'] = [x[2] for x in chip_extracted]

display(df[['Chipset', 'CPU', 'Hãng Chip', 'Số Nhân CPU', 'Xung Nhịp CPU']].head(10))
```

**Ở các cột có đơn vị, em sẽ tiến hành loại bỏ đơn vị đi**

```python
import pandas as pd
import numpy as np

# Tạo một bản sao để tránh cảnh báo SettingWithCopyWarning
df_features = df_features.copy()

# Thay vì tìm chính xác chuỗi ' GB', ta dùng Regex để xóa mọi ký tự đơn vị và khoảng trắng
# Chỉ giữ lại các chữ số (0-9) và dấu chấm (.) để chuyển đổi an toàn
unit_columns = [
    'RAM', 'ROM', 'Kích Thước Màn Hình', 'Tỷ Lệ Màn Hình', 
    'Camera Sau', 'Camera Trước', 'Dung Lượng Pin', 
    'Độ Dày', 'Trọng Lượng', 'Số Nhân CPU', 'Xung Nhịp CPU'
]

for col in unit_columns:
    if col in df_features.columns:
        # Xóa các ký tự đơn vị (kể cả '%' hay chữ tiếng Việt như 'nhân') và khoảng trắng bằng Regex
        df_features[col] = df_features[col].astype(str).str.replace(r'[^\d.]+', '', regex=True)
        # Chuyển về dạng số, các chuỗi 'nan' hoặc lỗi sẽ thành NaN
        df_features[col] = pd.to_numeric(df_features[col], errors='coerce')

# Đổi tên cột để người dùng vẫn biết đơn vị gốc sau khi bỏ chữ
rename_dict = {
    'RAM': 'RAM (GB)',
    'ROM': 'ROM (GB)',
    'Kích Thước Màn Hình': 'Kích Thước Màn Hình (inch)',
    'Tỷ Lệ Màn Hình': 'Tỷ Lệ Màn Hình (%)',
    'Camera Sau': 'Camera Sau (MP)',
    'Camera Trước': 'Camera Trước (MP)',
    'Dung Lượng Pin': 'Dung Lượng Pin (mAh)',
    'Độ Dày': 'Độ Dày (mm)',
    'Trọng Lượng': 'Trọng Lượng (g)',
    'Số Nhân CPU': 'Số Nhân CPU (Core)',
    'Xung Nhịp CPU': 'Xung Nhịp CPU (GHz)'
}
df_features = df_features.rename(columns=rename_dict)

# Hiển thị kết quả sau khi làm sạch đơn vị
display(df_features.head(15))

# Chỉ gọi info(), không cần bọc trong display()
df_features.info()
```

**Ở phần độ phân giải màn hình, em sẽ bỏ phần đơn vị đi, tính số pixel đường chéo, để sau này sẽ tiện tính PPI, tạo thành feature là số pixel**

```python
import re
import pandas as pd
import numpy as np

def calculate_diagonal_pixels(resolution_str):
    if pd.isna(resolution_str):
        return np.nan
    # Tìm các chuỗi số có dạng: số1 x số2 hoặc số1 X số2
    match = re.search(r'(\d+)\s*[xX]\s*(\d+)', str(resolution_str))
    if match:
        width = int(match.group(1))
        height = int(match.group(2))
        # Tính Pytago: căn bậc hai của (width^2 + height^2)
        return np.sqrt(width**2 + height**2)
    return np.nan

# Áp dụng hàm để tính số pixel trên đường chéo và bỏ cột chuỗi độ phân giải gốc
if 'Độ Phân Giải Màn Hình' in df_features.columns:
    df_features['Độ Phân Giải Đường Chéo (Pixels)'] = df_features['Độ Phân Giải Màn Hình'].apply(calculate_diagonal_pixels)
    df_features = df_features.drop(columns=['Độ Phân Giải Màn Hình'])

display(df_features[['Hãng Sản Xuất', 'Độ Phân Giải Đường Chéo (Pixels)']].head(10))
print(df_features['Độ Phân Giải Đường Chéo (Pixels)'].dtype)
```
